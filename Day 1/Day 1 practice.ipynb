```python
products = [
    ("iPhone", "Apple", 999),
    ("Galaxy", "Samsung", 799),
    ("MacBook", "Apple", 1299),
    ("iPad", "Apple", 899),
    ("Pixel", "Google", 699)
]

df = spark.createDataFrame(products, ["product", "brand", "price"])
df.show()
```

```python
df.select("product", "price").show()
```

```python
df.filter(df.price > 800).show()
```

```python
df.orderBy(df.price.desc()).show()
```

```python
from pyspark.sql.functions import col, round

df_with_tax = df.withColumn("price_with_tax",round(col("price") * 1.18,2))
df_with_tax.show()
```

```python
df.groupBy("brand").count().show()
```

```python
from pyspark.sql.functions import avg

df.groupBy("brand").agg(round(avg("price"), 2).alias("avg_price")).show()
```

```python
df.createOrReplaceTempView("products")

spark.sql("""
SELECT brand, COUNT(*) as total_products, round(AVG(price),2) as avg_price
FROM products
GROUP BY brand
""").show()
```

```python
df.write.mode("overwrite").saveAsTable("product_table")
```

```python
spark.sql("SELECT * FROM product_table").show()
```
